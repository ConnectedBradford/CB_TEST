```python
from google.cloud import bigquery

def get_all_dataset_sizes(project_id):
    client = bigquery.Client(project=project_id)
    datasets = client.list_datasets(project=project_id)

    summary = []

    print(f"Scanning datasets in project: {project_id}\n")

    for dataset in datasets:
        dataset_id = dataset.dataset_id
        dataset_ref = client.dataset(dataset_id)
        total_bytes = 0

        try:
            tables = client.list_tables(dataset_ref)
            for table in tables:
                table_ref = dataset_ref.table(table.table_id)
                table_obj = client.get_table(table_ref)
                if table_obj.table_type == "TABLE":
                    total_bytes += table_obj.num_bytes
        except Exception as e:
            print(f"⚠️  Skipping dataset `{dataset_id}` due to error: {e}")
            continue

        total_gb = round(total_bytes / 1024 / 1024 / 1024, 2)
        summary.append({
            "dataset_id": dataset_id,
            "size_gb": total_gb
        })

    # Print summary
    print("Dataset Sizes:")
    for d in summary:
        print(f"{d['dataset_id']:30} {d['size_gb']:10.2f} GB")

    return summary

if __name__ == "__main__":
    project_id = "yhcr-prd-bradfor-bia-core"  # Replace with your project ID
    get_all_dataset_sizes(project_id)
```
